# Parliament demographics

Ages are calculated at the parliament opening date, never at notebook execution time.

```python
# ruff: noqa: E402
import os
import sys
from pathlib import Path

root = Path(os.environ.get("APEMAP_PROJECT_ROOT", Path.cwd())).resolve()
while not (root / "pyproject.toml").exists() and root != root.parent:
    root = root.parent
sys.path.insert(0, str(root))

from apemap.analysis import compute_parliament_demographics
from apemap.db import get_connection

parliament = int(os.environ.get("APEMAP_PARLIAMENT", "48"))
db_path = Path(
    os.environ.get("APEMAP_DB_PATH", root / "data" / "aped.duckdb")
).resolve()
conn = get_connection(db_path, read_only=True)
demographics = compute_parliament_demographics(conn, parliament)
demographics
```

The age mean and median use `known_age_sample_size`; `missing_age_count` and the total parliamentarian count make the denominator explicit.
